```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Tạo dữ liệu giả lập
np.random.seed(42)
dates = pd.date_range("2026-01-01", periods=30)
values = np.random.randn(30).cumsum() + 100

df = pd.DataFrame({"Date": dates, "Score": values})
df.head()
```

```python
plt.figure(figsize=(9, 4))
plt.plot(df["Date"], df["Score"], marker="o", color="b", linestyle="-")
plt.title("Biểu đồ xu hướng chỉ số theo thời gian")
plt.xlabel("Ngày")
plt.ylabel("Giá trị")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
